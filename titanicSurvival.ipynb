```python
# Classification problem
# import libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = sns.load_dataset("titanic")
```

```python
df.head()
```

```python
# dropping same columns
df = df.drop(columns=['who','adult_male','alive','deck','class'])
```

```python
df.head()
```

```python
df.isnull().sum()
```

```python
df['embarked'].values
```

```python
df['embark_town'].values
```

```python
# dropping embarked
df = df.drop('embarked',axis=1)
```

```python
df.head()
```

```python
df['parch'].value_counts()
```

```python
df['embark_town'] = df['embark_town'].fillna('unknown')
```

```python
df.isnull().sum()
```

```python
df['age'] = df['age'].fillna(df['age'].mean())
```

```python
df.isnull().sum()
```

```python
# Label Encoding
from sklearn.preprocessing import LabelEncoder
```

```python
le = LabelEncoder()
```

```python
df['sex'] = le.fit_transform(df['sex'])
```

```python
df['embark_town'] = le.fit_transform(df['embark_town'])
```

```python
df.head()
```

```python
df['alone'] = le.fit_transform(df['alone'])
```

```python
df.head()
```

```python
# train_test_split
from sklearn.model_selection import train_test_split
X = df.drop('survived',axis=1)
y = df['survived']
```

```python
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.20,random_state=42)
```

```python
X_train
```

```python
print(X_train.shape,X_test.shape)
```

```python
# scaling the age and fare column
from sklearn.preprocessing import StandardScaler
```

```python
scaler = StandardScaler()
```

```python
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.fit_transform(X_test)
```

```python
X_train_scaled
```

```python
# baseline model
from sklearn.linear_model import LogisticRegression
import warnings
warnings.filterwarnings("ignore")
```

```python
model_lr = LogisticRegression()
```

```python
model_lr.fit(X_train,y_train)
```

```python
y_pred_lr = model_lr.predict(X_test)
```

```python
y_pred_lr
```

```python
# eavluate model
from sklearn.metrics import accuracy_score,classification_report,confusion_matrix
```

```python
accuracy = accuracy_score(y_test,y_pred_lr)
```

```python
accuracy
```

```python
# KNearestNeighbors
from sklearn.neighbors import KNeighborsClassifier
```

```python
model_knn = KNeighborsClassifier(n_neighbors=5)
```

```python
model_knn.fit(X_train_scaled,y_train)
```

```python
y_pred_knn = model_knn.predict(X_test_scaled)
```

```python
y_pred_knn
```

```python
# evaluate model
accuracy = accuracy_score(y_test,y_pred_knn)
```

```python
accuracy
```

```python
model_nb = GaussianNB()
```

```python
model_nb.fit(X_train,y_train)
```

```python
y_pred_nb = model_nb.predict(X_test)
```

```python
y_pred_nb
```

```python
# evaluate model
accuracy = accuracy_score(y_test,y_pred_nb)
```

```python
model_svm.fit(X_train_scaled,y_train)
```

```python
y_pred_svm = model_svm.predict(X_test_scaled)
```

```python
# evaluate model
accuracy = accuracy_score(y_test,y_pred_svm)
```

```python
# Decision Tree
from sklearn.tree import DecisionTreeClassifier
```

```python
y_pred_dt = model_dt.predict(X_test_scaled)
```

```python
# cross validation
from sklearn.model_selection import GridSearchCV
```

```python
param_grid = {
    'n_neighbors':[1,3,5,7,9,11,13,19],
    'weights':['uniform','distance'],
    'metric':['euclidean','manhattan'],
    'p':[1,2]
}

knn = KNeighborsClassifier()
```

```python
grid = GridSearchCV(
    estimator=knn,
    param_grid=param_grid,
    cv=5,
    scoring='accuracy',
    n_jobs=-1
)
```

```python
grid.fit(X_train_scaled,y_train)
```

```python
print("Best parameters: ",grid.best_params_)
print("Best Accuracy : ",grid.best_score_)
```

```python
# applying random forest
from sklearn.ensemble import RandomForestClassifier
```

```python
model_rf = RandomForestClassifier(random_state=42)
```

```python
model_rf.fit(X_train,y_train)
```

```python
y_pred_rf = model_rf.predict(X_test)
```

```python
y_pred_rf
```

```python
accuracy = accuracy_score(y_test,y_pred_rf)
```

```python
accuracy
```

```python
# hyperparameter tunning
param_grid = {
    'n_estimators':[100,200,300],
    'max_depth':[None,5,10,15],
    'min_samples_split':[2,5,10],
    'min_samples_leaf':[1,2,4],
    'max_features':['sqrt','log2']
}

grid = GridSearchCV(
    model_rf,
    param_grid,
    cv=5,
    scoring='accuracy',
    n_jobs=-1
)
grid.fit(X_train,y_train)
```

```python
print("Best parameters: ",grid.best_params_)
print("Best Accuracy : ",grid.best_score_)
```

```python
# Gradient Boosting
from sklearn.ensemble import GradientBoostingClassifier
model_gb = GradientBoostingClassifier(random_state=42)
model_gb.fit(X_train,y_train)
y_pred_gb = model_gb.predict(X_test)
print(accuracy_score(y_test,y_pred_gb))
```

```python
# hyper parameter tunning
param_grid = {
    'n_estimators':[100,200],
    'learning_rate':[0.01,0.05,0.1],
    'max_depth':[3,4,5],
    'subsample':[0.8,1.0]
}
```

```python
grid = GridSearchCV(
    model_gb,
    param_grid,
    cv=5,
    scoring='accuracy',
    n_jobs=-1
)
grid.fit(X_train,y_train)
```

```python
print("Best parameters: ",grid.best_params_)
print("Best Accuracy : ",grid.best_score_)
```

```python
# xgboost
from xgboost import XGBClassifier
```

```python
xgb = XGBClassifier(random_state=42,eval_metric='logloss')
```

```python
xgb.fit(X_train,y_train)
```

```python
y_pred = xgb.predict(X_test)
```

```python
print(accuracy_score(y_test,y_pred))
```

```python
param_grid = {
    'n_estimators':[100,200,300],
    'learning_rate':[0.01,0.05,0.1],
    'max_depth':[3,4,5],
    'subsample':[0.8,1.0],
    'colsample_bytree':[0.8,1.0]
}

grid = GridSearchCV(
    xgb,
    param_grid,
    cv=5,
    scoring='accuracy',
    n_jobs=-1
)
grid.fit(X_train,y_train)
```

```python
print("Best parameters: ",grid.best_params_)
print("Best Accuracy : ",grid.best_score_)
```
